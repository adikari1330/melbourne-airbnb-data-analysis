Which Melbourne neighbourhoods have the most listings?

```python
neighbourhood_counts = df["neighbourhood"].value_counts().head(15)

neighbourhood_counts
```

```python
plt.figure(figsize=(10, 6))

neighbourhood_counts.sort_values().plot(kind="barh")

plt.title("Top 15 Melbourne Neighbourhoods by Number of Airbnb Listings")
plt.xlabel("Number of Listings")
plt.ylabel("Neighbourhood")

plt.show()
```

Which neighbourhoods are the most expensive?

```python
neighbourhood_stats = (
    price_df.groupby("neighbourhood")
    .agg(
        average_price=("price", "mean"),
        median_price=("price", "median"),
        listing_count=("price", "count")
    )
)

neighbourhood_stats = neighbourhood_stats[
    neighbourhood_stats["listing_count"] >= 30
]

neighbourhood_stats.sort_values(
    "average_price",
    ascending=False
).head(15)
```

```python
top_expensive = neighbourhood_stats.sort_values(
    "average_price",
    ascending=False
).head(15)

plt.figure(figsize=(10, 6))

top_expensive["average_price"].sort_values().plot(kind="barh")

plt.title("Top Melbourne Neighbourhoods by Average Airbnb Price")
plt.xlabel("Average Price per Night ($)")
plt.ylabel("Neighbourhood")

plt.show()
```
